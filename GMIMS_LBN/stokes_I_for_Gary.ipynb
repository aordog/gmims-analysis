```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
from astropy.wcs import WCS
from astropy.coordinates import SkyCoord, ICRS, Galactic
import astropy.units as u
import matplotlib as mpl
from reproject import reproject_from_healpix, reproject_to_healpix
import healpy as hp
```

```python
freqs = np.genfromtxt('/srv/data/dragons/Mar2024_500_1000_MHz_cube/ListOfFrequencies_Hz.txt')/1e6
fidx = abs(freqs - 750).argmin()

print(freqs[fidx])
```

```python
hdu  = fits.open('/srv/data/dragons/Mar2024_500_1000_MHz_cube/DRAGONS_500_to_1024_i_cube_32bit.fits')
data = hdu[0].data[fidx]
print(data.shape)

hdr = hdu[0].header
#print(repr(hdr))
wcs = WCS(hdr).dropaxis(2)

hdr2D = wcs.to_header

print(repr(hdr2D))
```

```python
data[0:40] = np.nan
plt.imshow(data,origin='lower',vmin=-100,vmax=100)
```

```python
data_hpx, footprint = reproject_to_healpix((data,wcs),coord_system_out='C',nside=512)
```

```python
hp.mollview(data_hpx,cmap='magma',title='DRAGONS Stokes I, '+str(np.round(freqs[fidx],3))+' MHz',
                min=-400,max=400,xsize=3000, unit='Jy/beam')

plt.savefig('../PLOTS/dragons_stokesI_746.png',dpi=300)
```
