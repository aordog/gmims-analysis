```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
from astropy.wcs import WCS
#import healpy as hp
#from reproject import reproject_from_healpix, reproject_to_healpix
#import matplotlib.gridspec as gridspec
#import matplotlib.cm as cm
```

```python
hduQ = fits.open('/srv/data/chime/chime_IQUV_Mar2024_400_729/Q_400_729_Mar2024_new_gal_fix.fits')
hduU = fits.open('/srv/data/chime/chime_IQUV_Mar2024_400_729/U_400_729_Mar2024_new_gal_fix.fits')

Q = hduQ[0].data
U = hduU[0].data

freqs = WCS(hduQ[0].header).all_pix2world(0,0,range(Q.shape[0]),0)[2]
```

```python
df = 5e6
fpick = 600e6
idx = np.where(abs(fpick - freqs) < df/2)

print(freqs[idx])

Q600 = np.nanmedian(Q[idx], axis=0)
U600 = np.nanmedian(U[idx], axis=0)
```

```python
plt.imshow(Q600, origin='lower', vmin=-1, vmax=1)
```

```python
plt.imshow(U600, origin='lower', vmin=-1, vmax=1)
```

```python
hdr2D = WCS(hduQ[0].header).dropaxis(2).to_header()

print(repr(hdr2D))
```

```python
fits.writeto('/srv/data/chime/maps_for_Tom_Feb2026/Q600MHZ_BW5MHZ.fits', Q600, hdr2D, overwrite=False)
fits.writeto('/srv/data/chime/maps_for_Tom_Feb2026/U600MHZ_BW5MHZ.fits', U600, hdr2D, overwrite=False)
```
